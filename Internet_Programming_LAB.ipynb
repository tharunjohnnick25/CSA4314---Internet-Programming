```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(1)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(5)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(9)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(13)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(17)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(21)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(25)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(29)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(33)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(37)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(41)
```

```python
from IPython.display import HTML, display
import json, html as _html
PROJECTS = [('Library Management System', 'Books, study materials, borrowing history, fines', 'Search books and materials, view availability, borrow/return demo, calculate overdue fines.'), ('Hotel Room Booking', 'Room types, dates, guests, prices', 'Search rooms, check sample availability, create booking and view booking list.'), ('Bus Ticket Booking', 'Routes, schedules, seats, passengers', 'Search buses, select route, book sample ticket and view booking confirmation.'), ('Train Reservation System', 'Train routes, classes, seats, PNR', 'Search trains, select class, generate demo PNR, check booking status.'), ('Flight Reservation System', 'Flights, destinations, classes, e-tickets', 'Search sample flights, select class, book a flight and display e-ticket.'), ('Online Cake Ordering', 'Cakes, flavors, occasions, delivery', 'Filter cakes, add to cart, enter delivery address and place demo order.'), ('Beauty Parlour Appointment Booking', 'Services, stylists, appointment slots', 'Choose service and stylist, pick date/time, book or cancel appointment.'), ('Hospital Doctor Appointment', 'Hospitals, doctors, specialties, patient appointments', 'Search doctor by specialty, book slot, view appointment status.'), ('TechSavvy Electronics Store', 'Smartphones, laptops, tablets, accessories', 'Search/filter products, add to cart, update quantities and checkout.'), ('StyleHub Fashion Store', 'Clothing, footwear, accessories, sizes', 'Browse categories, filter products, add to cart and place demo order.'), ('Home Finder Real Estate', 'Houses, lands, rentals, agents', 'Filter properties by location and budget, save favorites and request a viewing.'), ('InsureEasy Insurance Management', 'Health, auto, home policies, agents', 'Compare sample policies, manage policy details and calculate reminder dates.'), ('BlossomGifts Flower and Gift Shop', 'Flowers, gift baskets, occasion gifts', 'Filter gifts, add to cart, schedule delivery and place demo order.'), ('ReNewMarket Second-hand Marketplace', 'Used electronics, furniture, clothing', 'Browse listings, search items, post a sample listing and mark item as sold.'), ('EasyToll Online Toll Payment', 'Toll passes, wallet, QR tickets, transactions', 'View wallet, calculate sample toll, purchase demo pass and display QR-like ID.'), ('ComplainEase Complaint Management', 'Complaints, officers, status, proof of work', 'Submit complaint, track status, officer updates and feedback.'), ('VoteOnline E-voting Demo', 'Candidates, voter verification, election results', 'Select a sample candidate and submit a demo vote; not for real elections.'), ('AgriMart Agricultural Marketplace', 'Seeds, fertilizers, tools, farming equipment', 'Search farm products, add to cart and view order/shipping status.'), ('VetCure Animal Telemedicine', 'Farmers, doctors, symptoms, consultations', 'Submit animal symptoms, attach text description, browse sample doctor responses.'), ('RentRide Vehicle Rental', 'Cars, trucks, SUVs, rental bookings', 'Search vehicles, select rental dates, submit booking and view status.'), ('Online Job Portal', 'Vacancies, directors, applicants, resumes', 'Search vacancies by category/location and submit a sample application.'), ('Solid Waste Management', 'Bins, trucks, drivers, routes, complaints', 'Report a bin issue, view collection tasks and simulate route assignment.'), ('Blood Donor Management', 'Donors, blood groups, hospitals, blood banks', 'Search donors/blood banks by group and location; register a donor demo.'), ('Wanderlust Travel Agency', 'Food, hotels, routes, destinations', 'Browse destination information and create a sample accommodation booking.'), ('Roadside Breakdown Assistance', 'Licensed mechanics, locations, feedback', 'Search mechanics by location and submit an assistance request.'), ('Traffic Violation FIR and Challan', 'Police cases, vehicle details, challans, proof', 'View sample challan and create a demo traffic case with violation details.'), ('Bus Pass and QR Ticket', 'Student/route passes, conductors, QR validation', 'Apply/renew bus pass, view sample pass ID and simulate conductor verification.'), ('ArtistryHub Artist Portfolio', 'Artist biography, artworks, gallery, events', 'Browse portfolio items, open artwork details and send an inquiry.'), ('Event Information and Registration', 'Speakers, venue, schedule, audience, registration', 'View event schedule and submit a sample registration.'), ('IJSSE Journal Publishing', 'Authors, reviewers, editors, articles, manuscripts', 'Browse journal sections and submit a demo manuscript entry.'), ('Ko-Matha Dairy Farm', 'Milk products, stock, customers, vendors', 'Browse dairy products, create a sample order and view stock report.'), ('Ram Infotech Laptop Sales and Service', 'Laptop models, repairs, service tracking', 'Browse laptops, request a repair and track a sample service request.'), ('KodaikanalEats Restaurant/Room Booking', 'Listings, room types, availability, prices', 'Search sample listings, check availability and book a sample slot.'), ('LearnWell Tuition Centre', 'Teachers, students, subjects, fees, schedules', 'Browse courses, register a student and view sample fee status.'), ('Bus Pass Management System', 'Routes, bus passes, e-tickets, conductors', 'Create pass application, renew pass and simulate QR validation.'), ('EduConnect Online Teaching', 'Courses, teachers, students, materials', 'Browse courses, enroll in a sample course and view study materials.'), ('BookHaven Online Bookstore', 'Books, genres, authors, reviews', 'Search books, filter by genre, add to cart and checkout.'), ('MyProfessionalShowcase Portfolio', 'Skills, projects, resume, blog, contact', 'Show portfolio projects, resume download link placeholder and contact form.'), ('Gourmet Haven Restaurant Website', 'Menu, dishes, table reservations, reviews', 'Browse menu, choose guests/date/time and submit a table reservation.'), ('Generic Online Shopping Store', 'Products, categories, cart, checkout', 'Search products, adjust cart quantities and generate a demo order summary.'), ('Travel Blog Website', 'Destinations, articles, maps, photo galleries', 'Filter travel posts by destination and submit a comment.'), ('Digital Recipe Book', 'Recipes, ingredients, cooking steps, meal planner', 'Search/filter recipes, scale ingredients and print recipe details.'), ('Electricity Bill Payment', 'Domestic/commercial units, due date, fine, receipt', 'Calculate electricity bill using supplied slabs and late fee.'), ('Placement Examination Portal', 'Exam schedule, questions, candidate results', 'Take a small demo quiz, calculate score and show result.'), ('Painting Rating and Review', 'Paintings, ratings, comments, owner packages', 'Rate a sample painting, submit review and calculate average rating.'), ('Income Tax Calculator', 'Gross salary, HRA, pension, insurance, tax slabs', 'Calculate taxable income and tax using the rules in the supplied brief.'), ('Phone Bill Payment', 'Monthly bill, due date, payment history, fine', 'Calculate overdue fine, enter payment amount and generate demo receipt.')]

def render_app(number):
    title, subtitle, features = PROJECTS[number-1]
    safe_title, safe_subtitle, safe_features = map(_html.escape, (title, subtitle, features))
    products = [{"name":"Starter Package","category":"General","price":499},{"name":"Standard Option","category":"Popular","price":899},{"name":"Premium Choice","category":"Premium","price":1499},{"name":"Budget Option","category":"Budget","price":299}]
    product_json = json.dumps(products)
    special = ""
    if number == 43:
        special = """<section class="panel"><h2>Electricity Bill Calculator</h2><label>Connection type <select id="conn"><option value="domestic">Domestic</option><option value="commercial">Commercial</option></select></label><label>Units consumed <input id="units" type="number" min="0" value="180"></label><label>Days after due date <input id="late" type="number" min="0" value="0"></label><button onclick="calcElectricity()">Calculate bill</button><p id="calcresult" class="result"></p></section>"""
    elif number == 46:
        special = """<section class="panel"><h2>Income Tax Calculator (lab-provided slabs)</h2><div class="formgrid"><label>Monthly gross pay (₹)<input id="gross" type="number" value="60000"></label><label>Monthly HRA (₹)<input id="hra" type="number" value="10000"></label><label>Monthly pension contribution (₹)<input id="pension" type="number" value="5000"></label><label>Monthly health insurance (₹)<input id="health" type="number" value="1000"></label></div><button onclick="calcTax()">Calculate tax</button><p id="taxresult" class="result"></p></section>"""
    elif number == 47:
        special = """<section class="panel"><h2>Phone Bill Payment Demo</h2><label>Outstanding bill (₹)<input id="bill" type="number" value="799"></label><label>Days after due date <input id="days" type="number" min="0" value="0"></label><label>Amount to pay (₹)<input id="payamt" type="number" value="799"></label><button onclick="payPhone()">Calculate and generate receipt</button><p id="phoneresult" class="result"></p></section>"""
    elif number == 45:
        special = """<section class="panel"><h2>Painting Review Demo</h2><label>Your rating <select id="rating"><option>5</option><option>4</option><option>3</option><option>2</option><option>1</option></select></label><label>Review <textarea id="review" placeholder="Write your feedback"></textarea></label><button onclick="addReview()">Submit review</button><p id="reviewresult" class="result"></p><ul id="reviews"></ul></section>"""
    elif number == 44:
        special = """<section class="panel"><h2>Sample Placement Quiz</h2><p>1. Which language structures a webpage? <select id="q1"><option value="">Choose</option><option>Python</option><option value="HTML">HTML</option><option>SQL</option></select></p><p>2. Which technology styles a webpage? <select id="q2"><option value="">Choose</option><option value="CSS">CSS</option><option>CSV</option><option>HTTP</option></select></p><button onclick="gradeQuiz()">Submit answers</button><p id="quizresult" class="result"></p></section>"""
    elif number == 42:
        special = """<section class="panel"><h2>Recipe and Ingredient Scaler</h2><label>Recipe <select id="recipe"><option value="2">Pancakes (2 servings)</option><option value="4">Vegetable Pasta (4 servings)</option></select></label><label>Desired servings <input id="servings" type="number" min="1" value="4"></label><button onclick="scaleRecipe()">Scale ingredients</button><p id="reciperesult" class="result"></p></section>"""
    else:
        special = """<section class="panel"><h2>Quick Action</h2><p>Try the form below to explore the main interaction for this project.</p><label>Search / item / service / reference <input id="searchbox" placeholder="Type a keyword"></label><label>Category <select id="category"><option>All</option><option>Popular</option><option>Budget</option><option>Premium</option></select></label><button onclick="runSearch()">Search</button><p id="actionresult" class="result">Ready. This demo uses sample data.</p><div id="items" class="items"></div><label>Name <input id="name" placeholder="Enter your name"></label><label>Details <textarea id="details" placeholder="Enter booking, request, feedback, or notes"></textarea></label><button onclick="submitDemo()">Submit demo request</button><p id="submitresult" class="result"></p></section>"""
    js = r"""
    const data = __PRODUCTS__;
    function runSearch(){const q=(document.getElementById('searchbox')?.value||'').toLowerCase();const c=document.getElementById('category')?.value||'All';const found=data.filter(x=>(!q||x.name.toLowerCase().includes(q)||x.category.toLowerCase().includes(q))&&(c==='All'||x.category===c));const el=document.getElementById('items');if(el)el.innerHTML=found.map(x=>`<article class="item"><b>${x.name}</b><span>${x.category}</span><strong>₹${x.price}</strong><button onclick="addItem('${x.name}',${x.price})">Select</button></article>`).join('')||'<p>No sample results found.</p>';const r=document.getElementById('actionresult');if(r)r.textContent=found.length+' sample result(s).';}
    let cart=[];function addItem(n,p){cart.push({n,p});const r=document.getElementById('actionresult');if(r)r.textContent=n+' selected. Items in cart: '+cart.length+'. Total: ₹'+cart.reduce((s,x)=>s+x.p,0);}
    function submitDemo(){const n=document.getElementById('name')?.value||'Guest';const d=document.getElementById('details')?.value||'No extra details';document.getElementById('submitresult').textContent='Demo submitted for '+n+'. Reference: DEMO-'+Math.random().toString(36).slice(2,8).toUpperCase()+'. Details: '+d;}
    function calcElectricity(){const u=Math.max(0,Number(document.getElementById('units').value)||0),type=document.getElementById('conn').value,days=Math.max(0,Number(document.getElementById('late').value)||0);let bill=0,remaining=u;const rates=type==='domestic'?[1,2.5,4,6]:[2,4.5,6,7],limits=[100,100,300,Infinity];for(let i=0;i<4&&remaining>0;i++){let chunk=Math.min(remaining,limits[i]);bill+=chunk*rates[i];remaining-=chunk;}const fine=days*10;document.getElementById('calcresult').textContent='Energy charge: ₹'+bill.toFixed(2)+' | Late fee: ₹'+fine.toFixed(2)+' | Total: ₹'+(bill+fine).toFixed(2)+'. Based on lab brief slabs.';}
    function calcTax(){const gross=Number(document.getElementById('gross').value)||0,hra=Number(document.getElementById('hra').value)||0,pension=Math.min(150000,Math.max(0,(Number(document.getElementById('pension').value)||0)*12)),health=Math.max(0,(Number(document.getElementById('health').value)||0)*12);const taxable=Math.max(0,gross*12-hra*12-50000-pension-health);let tax=0;if(taxable>1000000)tax=112500+.30*(taxable-1000000);else if(taxable>500000)tax=12500+.20*(taxable-500000);else if(taxable>250000)tax=.05*(taxable-250000);document.getElementById('taxresult').textContent='Net taxable income: ₹'+taxable.toFixed(2)+' | Estimated tax: ₹'+tax.toFixed(2)+' | After-tax annual income: ₹'+(gross*12-tax).toFixed(2)+'. Educational demo; verify current laws before filing.';}
    function payPhone(){const bill=Math.max(0,Number(document.getElementById('bill').value)||0),days=Math.max(0,Number(document.getElementById('days').value)||0),amount=Math.max(0,Number(document.getElementById('payamt').value)||0),fine=days*10;document.getElementById('phoneresult').textContent='Bill: ₹'+bill.toFixed(2)+' | Fine: ₹'+fine.toFixed(2)+' | Amount entered: ₹'+amount.toFixed(2)+' | '+(amount>=bill+fine?'Demo payment accepted':'Remaining: ₹'+(bill+fine-amount).toFixed(2))+' | Receipt DEMO-'+Date.now().toString().slice(-6);}
    function addReview(){const r=Number(document.getElementById('rating').value),t=document.getElementById('review').value||'No comment';let li=document.createElement('li');li.textContent=r+'/5 — '+t;document.getElementById('reviews').prepend(li);document.getElementById('reviewresult').textContent='Review added. Demo average based on this review: '+r+'/5.';}
    function gradeQuiz(){let score=(document.getElementById('q1').value==='HTML'?1:0)+(document.getElementById('q2').value==='CSS'?1:0);document.getElementById('quizresult').textContent='Score: '+score+'/2 ('+(score*50)+'%).';}
    function scaleRecipe(){const base=Number(document.getElementById('recipe').value),n=Math.max(1,Number(document.getElementById('servings').value)||1),ratio=n/base;document.getElementById('reciperesult').textContent='For '+n+' servings, use '+(2*ratio).toFixed(1)+' cups flour, '+(1*ratio).toFixed(1)+' cups milk and '+(1*ratio).toFixed(1)+' eggs (illustrative quantities).';}
    runSearch();
    """.replace("__PRODUCTS__", product_json)
    page = f"""<!doctype html><html><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1"><style>
    :root{{font-family:Arial,sans-serif;color:#182235;background:#f3f6fb}}*{{box-sizing:border-box}}body{{margin:0;padding:20px}}
    .hero{{background:linear-gradient(120deg,#172554,#4f46e5);color:white;padding:28px;border-radius:18px;margin:0 auto 18px;max-width:1000px}}
    .hero h1{{margin:0 0 8px;font-size:clamp(24px,4vw,36px)}}.hero p{{margin:0;line-height:1.5}}main{{max-width:1000px;margin:auto}}
    .panel{{background:white;padding:22px;border-radius:16px;box-shadow:0 5px 20px #1e293b12;margin:16px 0}}h2{{margin-top:0;color:#27356b}}
    label{{display:block;font-weight:600;margin:12px 0;color:#344054}}input,select,textarea{{display:block;width:100%;padding:11px;margin-top:6px;border:1px solid #cbd5e1;border-radius:9px;font:inherit;background:white}}textarea{{min-height:75px}}
    button{{border:0;border-radius:9px;padding:11px 16px;background:#4f46e5;color:white;font-weight:700;cursor:pointer;margin:6px 4px 6px 0}}button:hover{{background:#3730a3}}
    .items{{display:grid;grid-template-columns:repeat(auto-fit,minmax(180px,1fr));gap:12px;margin:14px 0}}.item{{background:#f8fafc;border:1px solid #e2e8f0;padding:14px;border-radius:12px;display:grid;gap:8px}}.item span{{color:#64748b}}.item strong{{font-size:18px}}
    .result{{background:#eef2ff;padding:12px;border-radius:10px;line-height:1.5;overflow-wrap:anywhere}}.features{{background:#e0f2fe;color:#0c4a6e;padding:12px 14px;border-radius:10px;line-height:1.6}}
    .formgrid{{display:grid;grid-template-columns:repeat(auto-fit,minmax(200px,1fr));gap:8px}}footer{{text-align:center;color:#64748b;padding:16px;font-size:12px}}</style></head><body><main>
    <header class="hero"><h1>{safe_title}</h1><p>{safe_subtitle}</p></header>
    <section class="panel"><h2>Project Features</h2><p class="features">{safe_features}</p><p><b>Experiment:</b> {number} of {len(PROJECTS)} · Interactive front-end prototype</p></section>
    {special}<footer>Internet Programming Lab · Sample demo only — no real transactions or personal data are stored.</footer></main><script>{js}</script></body></html>"""
    display(HTML(page))


# Run this experiment independently.
render_app(45)
```
